```python

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, binom

# Step 1: Create the dataset
marks = np.array([55, 60, 65, 68, 70, 72, 75, 78, 80, 85])

print("Student Marks Dataset:", marks)

# Step 2: Calculate mean and standard deviation
mean = np.mean(marks)
std = np.std(marks)

print("Mean:", round(mean, 2))
print("Standard Deviation:", round(std, 2))

# Step 3: Calculate normal distribution probabilities
x = np.linspace(40, 100, 200)
y = norm.pdf(x, mean, std)

# Probability of scoring above 75
probability = 1 - norm.cdf(75, mean, std)
print("Probability of marks above 75:", round(probability, 4))

# Step 4: Visualize the normal distribution
plt.figure(figsize=(8, 5))
plt.hist(marks, bins=6, density=True, alpha=0.5,
         edgecolor="black", label="Student Marks")
plt.plot(x, y, linewidth=2, label="Normal Distribution")
plt.xlabel("Marks")
plt.ylabel("Probability Density")
plt.title("Normal Distribution of Student Marks")
plt.legend()
plt.grid(True)
plt.show()

# Step 5: Bayes' Theorem
# Example: Medical test
prior = 0.01             # 1% chance of having a disease
sensitivity = 0.95       # Positive test if disease is present
specificity = 0.90       # Negative test if disease is absent

p_positive = (
    sensitivity * prior
    + (1 - specificity) * (1 - prior)
)

posterior = (sensitivity * prior) / p_positive

print("\nBayes' Theorem")
print("Probability of a positive test:", round(p_positive, 4))
print("Probability of disease given positive test:",
      round(posterior, 4))
print("Percentage:", round(posterior * 100, 2), "%")

# Step 6: Binomial distribution
n = 10
p = 0.5
k = np.arange(0, 11)
binomial_prob = binom.pmf(k, n, p)

plt.figure(figsize=(8, 5))
plt.bar(k, binomial_prob, edgecolor="black")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.title("Binomial Probability Distribution")
plt.xticks(k)
plt.grid(axis="y")
plt.show()
```
