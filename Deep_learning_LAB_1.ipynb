```python
import numpy as np
import pandas as pd
data = pd.read_csv("/content/student_performance_100.csv")
X = data[["Attendance", "Assignment_Score", "Internal_Marks"]].values
X = X / 100.0
y = data["Final_Marks"].values.reshape(-1, 1) / 100.0
np.random.seed(42)
W = np.random.randn(3, 1) * 0.1
b = 0.0
learning_rate = 0.1
epochs = 3000
n = len(X)
for epoch in range(epochs):
    y_pred = np.dot(X, W) + b
    error = y_pred - y
    loss = np.mean(error ** 2)
    dW = (2 / n) * np.dot(X.T, error)
    db = 2 * np.mean(error)
    W -= learning_rate * dW
    b -= learning_rate * db
predictions = np.clip((np.dot(X, W) + b) * 100, 0, 100)
print("Dataset Shape:", data.shape)
print("Input Matrix Shape:", X.shape)
print("Weight Matrix Shape:", W.shape)
print("Bias:", b)
print("Output Matrix Shape:", predictions.shape)
print("Final Loss:", loss)
print("\nFirst 10 Student Predictions:")
for i in range(10):
    print(
        "Student", i + 1,
        "| Actual Marks:", data["Final_Marks"].iloc[i],
        "| Predicted Marks:", round(float(predictions[i, 0]), 2)
    )
data["Predicted_Marks"] = predictions.flatten()
data.to_csv("student_performance_predictions.csv", index=False)
print("\nPredictions saved successfully!")
```
